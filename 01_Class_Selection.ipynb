```python
import os
import pandas as pd
```

Step 1 — Inspect the dataset

```python
# ==============================
# 2. Set dataset path
# ==============================

DATASET_PATH = "/content/drive/MyDrive/Colab Notebooks/Y 02 S 02, Advance ML/Assignment/MIT Indoor Scenes/indoorCVPR_09/Images"

print("Dataset exists:", os.path.exists(DATASET_PATH))
print("Dataset path:", DATASET_PATH)
```

```python
# ==============================
# 3. Inspect folders
# ==============================

original_image_counts = {}

for class_folder in os.listdir(DATASET_PATH):
    class_path_full = os.path.join(DATASET_PATH, class_folder)

    if os.path.isdir(class_path_full):
        images_in_folder = [f for f in os.listdir(class_path_full) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
        original_image_counts[class_folder] = len(images_in_folder)


sorted_image_counts = sorted(original_image_counts.items(), key=lambda item: item[1], reverse=True)

print("Types of places sorted by image count (descending):")
for class_name, count in sorted_image_counts:
    print(f"- {class_name}: {count} images")
```

```python
# ==============================
# 4. Count images in each class
# ==============================

class_counts = {}

for class_name in sorted(os.listdir(DATASET_PATH)):
    class_path = os.path.join(DATASET_PATH, class_name)

    if os.path.isdir(class_path):
        image_count = 0

        for file in os.listdir(class_path):
            if file.lower().endswith(('.jpg', '.jpeg', '.png')):
                image_count += 1

        class_counts[class_name] = image_count

class_df = pd.DataFrame(
    list(class_counts.items()),
    columns=["Class", "Image_Count"]
)

class_df = class_df.sort_values("Image_Count", ascending=False)

class_df
```

# Smart Home Indoor Scene Recognition from Images
## Notebook 01: Class Selection

### Project Overview

This notebook begins the image classification project for the HomeVision smart-home scenario. The objective is to develop an image classification system that can recognize different indoor environments from images.

The MIT Indoor Scenes dataset is used as the source dataset.

These classes represent different indoor environments that could be relevant to a smart-home or indoor scene recognition system. The selected classes also contain visually similar environments, such as bedrooms and living rooms or kitchens and bars, which provides a meaningful classification task.

### Main Objectives

1. Identify the available indoor scene classes in the dataset.
2. Select five suitable classes for the HomeVision scenario.
3. Inspect the number of images available for each selected class.
4. Record the selected classes and their image counts.
5. Use the selected classes consistently throughout the remaining notebooks.

### Dataset

The dataset used in this project is the MIT Indoor Scenes dataset (IndoorCVPR 2009). The original dataset is located in the project Google Drive directory.

## Step 1 - Import

```python
import os
import pandas as pd
from pathlib import Path
import kagglehub
```

## Step 2 — Set the Dataset Path

```python
%pip install -q -U kagglehub
```

```python
DATASET_PATH = kagglehub.dataset_download("itsahmad/indoor-scenes-cvpr-2019")

images_dir = Path(DATASET_PATH) / "indoorCVPR_09" / "Images"

print("Dataset path:", DATASET_PATH)
print("Images path:", images_dir)
print("Images folder exists:", images_dir.exists())
```

## Step 3 — Inspect the Dataset Folders

Each folder in the dataset is inspected and the number of valid image files is counted.

This provides an initial view of how many images are available for each indoor scene class.

```python
# ==============================
# 3. Inspect folders
# ==============================

original_image_counts = {}

for class_folder in os.listdir(images_dir):
    class_path_full = os.path.join(images_dir, class_folder)

    if os.path.isdir(class_path_full):
        images_in_folder = [f for f in os.listdir(class_path_full) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
        original_image_counts[class_folder] = len(images_in_folder)


sorted_image_counts = sorted(original_image_counts.items(), key=lambda item: item[1], reverse=True)

print("Types of places sorted by image count (descending):")
for class_name, count in sorted_image_counts:
    print(f"- {class_name}: {count} images")
```

## Step 4 — Count Images in Each Class

The number of JPG, JPEG and PNG images is calculated for every available class.

The resulting table is sorted by image count to make it easier to identify classes with sufficient images for the experiment.

```python
# ==============================
# 4. Count images in each class
# ==============================

class_counts = {}

for class_name in sorted(os.listdir(images_dir)):
    class_path = os.path.join(images_dir, class_name)

    if os.path.isdir(class_path):
        image_count = 0

        for file in os.listdir(class_path):
            if file.lower().endswith(('.jpg', '.jpeg', '.png')):
                image_count += 1

        class_counts[class_name] = image_count

class_df = pd.DataFrame(
    list(class_counts.items()),
    columns=["Class", "Image_Count"]
)

class_df = class_df.sort_values("Image_Count", ascending=False)

class_df
```

```python
# Classes with approximately 600 images

class_df[
    (class_df["Image_Count"] >= 500) &
    (class_df["Image_Count"] <= 750)
]
```
